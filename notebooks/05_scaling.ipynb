```python
import pandas as pd
import numpy as np

rows = []
for site_id, g in results.groupby('site_id'):
    actual = g['consumed_tonnes']
    forecast = g['forecasted_consumption']
    mask = actual != 0
    mape = (abs(actual[mask] - forecast[mask]) / actual[mask]).mean() * 100
    rmse = np.sqrt(((actual - forecast) ** 2).mean())
    rows.append({'site_id': site_id, 'MAPE': round(mape, 2), 'RMSE': round(rmse, 2)})

site_scores = pd.DataFrame(rows)
site_scores
```

```python
print('Average MAPE:', round(site_scores['MAPE'].mean(), 2), '%')
print('Sites meeting the 15% target:', (site_scores['MAPE'] <= 15).sum(), 'of 30')
```
